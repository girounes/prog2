```python
import pandas as pd

df = pd.read_csv('REL_ESTOQUE_31.08.2026_Grupo9.csv',
                 sep=';',
                 encoding='latin1')

print(df.head())
print(df.shape)
```

```python

# 9.2 - Exclusão de colunas

df = df.drop(columns=['Volumes,,,,,,,'])

print(df.head())
print(df.shape)
```

```python

# 9.3 - Renomeação das colunas

df = df.rename(columns={
    'Codigo': 'codigo_produto',
    'Descricao': 'descricao_produto',
    'Categoria': 'categoria',
    'Unidade': 'unidade_medida',
    'Filial': 'filial',
    'TipoEstoque': 'tipo_estoque',
    'Fornecedor': 'fornecedor',
    'EstoqueFisico': 'estoque_fisico',
    'EstoqueMinimo': 'estoque_minimo',
    'QtdEmbalagem': 'qtd_embalagem',
    'CustoUnitario': 'custo_unitario',
    'PrecoVenda': 'preco_venda',
    'UltimaCompra': 'data_ultima_compra',
    'QtdUltimaCompra': 'qtd_ultima_compra'
})

print(df.columns)
print(df.shape)
```

```python

# 9.4 - Identificação dos tipos de dados

df.info()
```

```python

# 9.4 - Conversão dos tipos de dados

colunas_numericas = [
    'estoque_fisico',
    'qtd_embalagem',
    'custo_unitario',
    'preco_venda',
    'qtd_ultima_compra'
]

for coluna in colunas_numericas:
    df[coluna] = df[coluna].str.replace(',', '.')
    df[coluna] = pd.to_numeric(df[coluna], errors='coerce')

# Verificação dos tipos após a conversão
df.info()
```

```python

# 9.5 - Tratamento das datas

# Visualização antes do tratamento
print('ANTES:')
print(df['data_ultima_compra'].head(10))

# Função para padronizar os formatos
def padronizar_data(data):
    if pd.isna(data):
        return data

    data = str(data).strip()

    if data[:4].isdigit():
        data = pd.to_datetime(data, format='%Y-%m-%d', errors='coerce')
    else:
        data = data.replace('-', '/')
        data = pd.to_datetime(data, format='%d/%m/%Y', errors='coerce')

    if pd.isna(data):
        return None

    return data.strftime('%d/%m/%Y')

# Padronização das datas
df['data_ultima_compra'] = df['data_ultima_compra'].apply(padronizar_data)

# Conversão para datetime
df['data_ultima_compra'] = pd.to_datetime(
    df['data_ultima_compra'],
    format='%d/%m/%Y',
    errors='coerce'
)

# Visualização após o tratamento
print('\nDEPOIS:')
print(df['data_ultima_compra'].head(10))

print('\nTipo da coluna:')
print(df['data_ultima_compra'].dtype)
```

```python

# 9.5 - Verificação das datas NAO NECESSARIO, APENAS PARA TESTES DE COERENCIA

print('Datas ausentes:', df['data_ultima_compra'].isna().sum())

data_base = pd.to_datetime('31/08/2026', format='%d/%m/%Y')

print('Compras posteriores à data-base:')
print(df[df['data_ultima_compra'] > data_base])
```

```python


# 9.6 - Identificação dos dados ausentes

print('Quantidade de dados ausentes por coluna:')
print(df.isnull().sum())

print('\nPercentual de dados ausentes por coluna:')
print((df.isnull().sum() / len(df) * 100).round(2))

```

```python

# 9.7 - Padronização dos valores

colunas = ['categoria', 'filial', 'tipo_estoque']

# Visualização antes da padronização
for coluna in colunas:
    print('\nANTES -', coluna)
    print(df[coluna].value_counts(dropna=False))

# Padronização dos valores
for coluna in colunas:
    df[coluna] = df[coluna].str.strip().str.title()

# Visualização após a padronização
for coluna in colunas:
    print('\nDEPOIS -', coluna)
    print(df[coluna].value_counts(dropna=False))
```

```python

# 9.6 - Tratamento dos dados ausentes

colunas = ['custo_unitario', 'preco_venda']

# Preenchimento dos valores ausentes pela mediana da categoria
for coluna in colunas:
    df[coluna] = df[coluna].fillna(
        df.groupby('categoria')[coluna].transform('median')
    )

# Verificação após o tratamento
print(df.isnull().sum())
```

```python

# 9.8 - Criação de variáveis numéricas e categóricas

import numpy as np

# Valor financeiro do estoque
df['valor_estoque'] = df['estoque_fisico'] * df['custo_unitario']

# Margem bruta percentual
df['margem_percentual'] = (
    (df['preco_venda'] - df['custo_unitario']) /
    df['preco_venda'] * 100
)

# Situação do estoque
df['situacao_estoque'] = np.where(
    df['estoque_fisico'].isna(),
    'Não informado',
    np.where(
        df['estoque_fisico'] < df['estoque_minimo'],
        'Abaixo do mínimo',
        'No mínimo ou acima'
    )
)

# Visualização das novas variáveis
print(df[['valor_estoque', 'margem_percentual',
          'situacao_estoque']].head(10))

print(df.shape)
```

```python

# 9.9 - Discretização da margem percentual

df['faixa_margem'] = pd.qcut(
    df['margem_percentual'],
    q=3,
    labels=['Baixa', 'Média', 'Alta']
)

# Visualização da nova variável
print(df[['margem_percentual', 'faixa_margem']].head(10))

# Quantidade de produtos em cada faixa
print(df['faixa_margem'].value_counts())
```

```python

# 9.10 - Criação de variáveis dummy

# Visualização antes da transformação
print('ANTES:')
print(df['tipo_estoque'].head(10))

# Criação das variáveis dummy
dummy_estoque = pd.get_dummies(
    df['tipo_estoque'],
    dtype=int
)

# Inclusão das variáveis no DataFrame
df[['Próprio', 'Consignado']] = dummy_estoque[['Próprio', 'Consignado']]

# Visualização após a transformação
print('\nDEPOIS:')
print(df[['tipo_estoque', 'Próprio', 'Consignado']].head(10))

print(df.shape)
```

```python

# 9.11 - Estrutura final do DataFrame

print('Estrutura final:')
df.info()

print('\nQuantidade de linhas e colunas:')
print(df.shape)

print('\nNomes das colunas:')
print(df.columns)
```
